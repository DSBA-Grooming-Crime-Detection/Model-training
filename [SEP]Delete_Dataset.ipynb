```python
from google.colab import files
import pandas as pd
import re
# 파일 업로드
uploaded = files.upload()

# 데이터프레임으로 파일 읽기
df = pd.read_csv('/content/cleaned_labeling.csv')

#" [SEP] "삭제
# [SEP] 및 양옆 공백 제거
df['text'] = df['text'].str.replace(r'\s*\[SEP\]\s*', ' ', regex=True)

# 로컬에 데이터프레임을 csv 파일로 다운로드
df.to_csv('cleaned_labeling_[SEP]Delete.csv', index=False)
files.download('cleaned_labeling_[SEP]Delete.csv')
```

```python
# sample 데이터 생성
# label 값이 '0'인 행에서 10개 샘플링
df_0 = df[df['label'] == 0].sample(n=10, random_state=42)

# label 값이 '1'인 행에서 10개 샘플링
df_1 = df[df['label'] == 1].sample(n=10, random_state=42)

# 두 개 합치기
df_sample = pd.concat([df_0, df_1]).reset_index(drop=True)

# 로컬에 데이터프레임을 csv 파일로 다운로드
df_sample.to_csv('cleaned_labeling_[SEP]Delete_sample.csv', index=False)
files.download('cleaned_labeling_[SEP]Delete_sample.csv')
```
